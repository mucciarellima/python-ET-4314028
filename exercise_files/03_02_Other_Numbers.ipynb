```python
from decimal import Decimal, getcontext #from decimal module import Decimal class and getcontext function
```

## Integers

```python
int('100') #casting string 100 to integer 100
```

```python
int('100', 2) #second argument says that base is 2 ie. for binary conversion (base 10 is default)
#100 in binary is 4+0+0=4 in decimal
```

```python
int(100, 2)#in order to convert integer from binary to decimal, first arg must be a string
```

```python
int('1ab', 16) #hexadecimal to decimal conversion
```

```python
1.2 - 1.0 # results in 0.19999999999999996 due to floating point precision issues (floating point error)
```
