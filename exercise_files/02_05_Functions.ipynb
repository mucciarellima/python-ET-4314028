<span style="color: yellow;">for the purposes here, FACTORIAL is defined for positive integers and 0 only (negative or decimal numbers should return NONE)

```python
def factorial(num):
    # Your code goes here.
    s=num
    s1=0
    answer=None
    if type(num) == int and num >= 0:
        if num == 0 :
            answer = 1
        else:
            answer=num
            s1=num-1
            while s > 1:
                answer=answer*s1
                s=s-1
                s1=s1-1
    return answer
```

```python
def factorialB(num):
    answer = None
    if type(num) == int and num >= 0:
        if num == 0:
            answer = 1
        else:
            answer = 1
            for i in range(1, num + 1):
                answer = answer * i
    return answer
```

```python
print(factorial(3.5))
```

```python
print(factorialB(3.5))
```

```python
print("Testing factorial with different numbers:")
for n in ["spam",-1,0, 1, 2, 3, 4, 5,16,121.5,.44]:
    print(f"factorial({n}) = {factorial(n)}")
```

```python
print("Testing the new factorial function with different numbers:")
for n in ["spam",-1,0, 1, 2, 3, 4, 5,16,211.5]:
    print(f"factorial({n}) = {factorialB(n)}")
```

<span style="color: yellow;"><span style="color: yellow;">here is how you do it using RECURSION

```python
def factorialRecursion(num):
    # Your code goes here.
    if type(num) != int:
        return None
    if num < 0:
        return None
    if num == 0:
        return 1

    return num * factorialRecursion(num-1)
```

```python
print("Testing the new factorial function with different numbers:")
for n in ["spam",-1,0, 1, 2, 3, 4, 5,16,21.5]:
    print(f"factorial({n}) = {factorialRecursion(n)}")
```
