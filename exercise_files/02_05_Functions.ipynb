## Functions

```python
print('Hello, World!')
```

```python
def multiplyByThree(val):
    return 3 * val

multiplyByThree(4)
```

```python
def multiply(val1, val2):
    print("Multiplying", val1, "and", val2,"and returning the result")
    return val1 * val2
```

```python
multiply(3, 4)
```

```python
x=multiply(5, 6)
#in this case, x will be assigned the return value of multiply(5,6)
print("printing the value of x....", x)

multiply(6,7)
#in this case, the return value of multiply(6,7) is not assigned to any variable and will be discarded
```

```python
a = [1,2,3]

def appendFour(myList):
    myList.append(4)
    print('peforming the function. now adding 4 to the list')
    
appendFour(a)
print(a)  # a is now [1,2,3,4]
```

in this next section, we are printing the returned value of the PRINT function. the PRINT function does not return anything so the result is NONE

```python
print(print('Hello, World!'))
```

```python
a = [1,2,3]

def appendFour(myList):
    myList.append(4)
    print('peforming the function. now adding 4 to the list but not returning anything')
    
appendFour(a)
print(a)  # a is now [1,2,3,4]
print(appendFour(a))  # this will print None because the function does not return anything
```
