## Strings

### Slicing

```python
name = 'My name is Ryan Mitchell'
```

```python
name[0] #m is at index 0 in the string
```

```python
name[1]
```

```python
name[0:7] #get from index 0 to index 7 (not including index 7)
```

```python
name[:7]#shorthand for name[0:7]
```

```python
name[11:]#get from index 11 to the end of the string
#handy for long strings when you don't know the length
```

```python
myList = [1,2,3,4,5]
```

```python
myList[2:4]#slice from index 2 to index 4 (not including index 4)
```

```python
len(name)
```

```python
len(myList)
```

### Formatting

```python
'My number is: '+str(5)
```

```python
f'My number is: {5}'
```

```python
f'My number is: {5} and twice that is {2*5}'
```

```python
f'Pi is: {math.pi:.2f}'#need to import the math module for math.pi to work
```

```python
'Pi is: {}'.format(math.pi)#old way
```
