### Boolean Logic

```python
weatherIsNice = True
haveUmbrella = False

if not (haveUmbrella or weatherIsNice): #parantheses are important here to get result we want
    #if weather is not nice and we don't have an umbrella then stay inside
    #ie. only if both conditions are False will the condition evaluate to True (because of the NOT operator)
    # the or operator returns false only if both operands are false (in this context it would mean that the weather is not nice and we don't have an umbrella)
    print('Stay inside')
else:
    print('Go for a walk')
```

```python
weatherIsNice = True
haveUmbrella = False

if (not haveUmbrella) and (not weatherIsNice):
    print('Stay inside')
else:
    print('Go for a walk')
```

```python
weatherIsNice = True
haveUmbrella = False

if haveUmbrella or weatherIsNice:
    print('Go for a walk')
else:
    print('Stay inside')
```
