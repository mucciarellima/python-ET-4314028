```python
mySet = set({'a', 'b', 'c'}) #another way to create a set using the set() function
```

```python
mySet = set(('a', 'b', 'c')) #here we used a tuple to create the set
```

```python
mySet #shows that we passed in a tuple and got a set back
```

```python
myList = ['a', 'b', 'b', 'c', 'c']#this list has duplicate values
myList = list(set(myList)) #here we first convert the list to a set becuase we want it to remove duplicates (set(myList does that)) and then back to a list
myList #but the order coming back out may be different
```

## Tuples - most efficinet memory storage

```python
myTuple = ('a', 'b', 'c')#declared using parentheses
myTuple
```

```python
myTuple[0] #tuples are subscriptable (order matters)
```

```python
myTuple[0] = 'd' #this will give an error because tuples are immutable
```

```python
def returnsMultipleValues():
    return 1,2,3 #this function returns multiple values, which are actually packed into a tuple

type(returnsMultipleValues())
```

```python
myTuple = 1,2,3 #shows that you don't need parentheses to declare a tuple
```

```python
type(myTuple)
```

```python
a, b, c = returnsMultipleValues() #this is called unpacking a tuple into multiple variables
#each variable gets assigned the corresponding value from the tuple
```

```python
print(a)
print(b)
print(c)
```
