## Dictionary Comprehensions

```python
animalList = [('a', 'aardvark'), ('b', 'bear'), ('c', 'cat'), ('d', 'dog')]#a list of tuples
animals = {item[0]: item[1] for item in animalList}
animals
```

```python
animals = {key: value for key, value in animalList}#creates a dictionary from a list of tuples
animals
#note that in the above example, we unpack each tuple into key and value variables
#this works as long as each tuple has exactly two elements
#the number of arguments on the left side of the for must match the number of elements in each tuple
```

```python
animalList2 = [('a', 'aardvark'), ('b', 'bear'), ('c', 'cat'), ('d', 'dog', 'deer'), ('e', 'elephant')]#this will cause an error because one tuple has 3 elements
animals2 = {key: value for key, value in animalList2}
animals2
```

```python
animalList3 = [('a', 'aardvark'), ('b', 'bear'), ('c', 'cat'), ('d', 'dog', 'deer'), ('e', 'elephant')]
animals3 = {key: value for key, value, value2 in animalList3}
#this will also won't work because we have not all the tuples have 3 elements
animals3
```

```python
animals.items()#returns an OBJECT containing key-value pairs as a list of tuples
```

```python
list(animals.items())#converts the object returned by items() into a list of tuples
```

```python
[{'field1': key, 'field2': value} for key, value in animals.items()]
#this returns a list of dictionary objects created from the key-value pairs in the animals dictionary object
#each dictionary has two fields: 'letter field' and 'name field'
#each field is populated with the corresponding key or value from the animals dictionary
```
