### Bytes

```python
bytes(4) #create a bytes object that's 4 bytes long (8 bits per byte = 256 possible values per byte)
#Each byte is initialized to zero by default.
#each byte object is represented as b'\x00' where 00 is the hex value of the byte (base 16)
#bytes objects are immutable (like tupples)
```

```python
smileyBytes = bytes('🙄', 'utf-8') #emojis are encoded with utf-8
smileyBytes
```

```python
smileyBytes.decode('utf-8')#decode the bytes object back into a string
```

```python
smileyBytes = bytearray('🙄', 'utf-8') #bytearray objects are not immutable
```

```python
f'{smileyBytes} has type {type(smileyBytes)} which is an object that can be modified'
```

```python
#illustrates that the bytearray object can be modified
smileyBytes[3] = int('85', 16)#replacing the 4th byte (index 3) with hex value 85
```

```python
smileyBytes.decode('utf-8')
```
