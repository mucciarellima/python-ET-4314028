## Classes
<span style="color: yellow;">\_\_init__ is a special function all classess have.  
It's the initialiaztion function   
\_\_init__ runs every time an instance of the class is created

<span style="color: yellow;">self refers to the specific instance of the class after it's instantiated  

<span style="color: yellow;"> While not a technical requirement for your code to execute, using Capitalization for CLASS names will make it easier to identify classes from functions and variables in your code  

```python
class Dog: #this creates a Dog Object; note the capital D in Dog
    def __init__(self, name):
        self.name = name #this is an attribute
        self.legs = 4  #this is another attribute
        #so each time the class is instantiated, the dog will have a name (which gets passed in at call time) and 4 legs
    
    def speak(self): #this is a method - self allows speak to have access to the attributes of the class
        print(self.name + ' says: Bark!') #this is code the defines what the method does  
```

```python
my_dog = Dog('Rover')  #you know this is a class becuase it's capitalized; this creates an instance of the Dog class where the name attribute is 'Rover'
another_dog = Dog('Fluffy')  #this creates another instance of the Dog class where the name attribute is 'Fluffy'
```

```python
my_dog.speak()
```

```python
another_dog.speak()
```

```python
print(another_dog.name, 'has',another_dog.legs,'legs.')
```

```python
print(f'{another_dog.name} has {another_dog.legs} legs. But when {my_dog.name} shows up ...')
another_dog.speak()
```
