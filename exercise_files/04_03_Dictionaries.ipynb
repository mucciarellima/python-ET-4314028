```python
from collections import defaultdict
```

## Dictionaries

```python
animals = {
    'a': 'aardvark',
    'b': 'bear',
    'c': 'cat',#note the trailing comma, not required but good practice
}
animals
```

```python
animals['a']
```

```python
animals['d'] = 'dog'#adding a new key-value pair to the dictionary
```

```python
animals
```

```python
animals['a'] = 'antelope'#changing the value for key 'a'
animals
```

```python
animals.keys()#gets all the keys in the dictionary
#returns a dict_keys object not a list even though it looks like a list
```

```python
animals.values()#gets all the values in the dictionary
#returns a dict_values object not a list even though it looks like a list
```

```python
list(animals.keys())#converts the dict_keys object to a list
```

```python
animals['e']#e is not a key in the dictionary, will raise a KeyError
```

```python
#animals.get('e')#better way to get a value from a dictionary becuase it hanles missing keys gracefully
#print(animals.get('e'))#returns None if the key is not found
animals.get('e','elephant')#here we provide a default value 'elephant' to return if the key is not found
print(animals.get('e','elphant'))#returns elephant if the key is not found
```

```python
len(animals)
```

```python
animals = { #this is a dictionary where each value is a list of values - this is a dictionary of lists
    'a': ['aardvark', 'antelope'],
    'b': ['bear'],
    'c':['elephant'],
}
animals
```

```python
animals['b'].append('bison')#adding 'bison' to the list of animals under key 'b'
animals
```

```python
animals['c'] = ['cat']#adding a new key 'c' with a list containing 'cat' as its value
#however, if c already exists, this will overwrite the existing list 
#here elephant will be lost
animals
```

```python
if 'c' not in animals:#check if key 'c' exists in the dictionary
    animals['c'] = []#initialize an empty list for key 'c' if it does not exist
    
animals['c'].append('caribou')#now we can safely append 'caribou' to the list under key 'c'
animals
```

### The Default Dict - import from the collections package

```python
animals = defaultdict(list)#this establishes the type of object that should be returned by default for missing keys
```

```python
animals
```

```python
animals['e'].append('elephant')
animals
```

```python
animals['e'].append('emu')
animals
```

```python
animals['f']#returns an empty list because 'f' is not yet a key in the dictionary
```

```python
animals['g'].append('giraffe')#automatically creates a new list for key 'g' and appends 'giraffe' to it
animals#note that f is still empty but exists as a key in the dictionary now
```
