```python
my_set = {1,1,2,2}
len(my_set)
print(my_set)
```

<span style="color: yellow;">this is not a set, it's a list, and in a list ORDER matters so the following evaluates to false

```python
[1,2] == [2,1]
```

<span style="color: yellow;">but order does not matter for a set nor does repeating elements. So the following evaluates to true

```python
{1,2,3} == {3,2,1,1,1}
```

## Dictionaries

```python
my_dictionary = {
    'apple': 'A red fruit', 
    'bear': 'A scary animal'
}
```

```python
my_dictionary['apple']
```

<span style="color: yellow;">watch out!!! duplicate keys cause problems. in this case, apple is reassigned the value that's entered last!!! so when you try to get the value for the apple key, Python retrieves the last one it assigned to that key. so duplicating keys do not result in an error, but when accessing the dictionary you might get unexpected results and find that the original key was replaced by something unexpected

```python
my_dictionary = {
    'apple': 'A red fruit', 
    'bear': 'A scary animal',
    'apple': 'Sometimes a green fruit'
}
```

<span style="color: yellow;">see this runs just fine

```python
my_dictionary['apple']
```

<span style="color: yellow;">see the last apple was returned, and the first one was overwritten

```python
my_dictionary
```
