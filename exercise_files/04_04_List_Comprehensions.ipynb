## List Comprehensions

```python
myList = [1,2,3,4,5]#creates a list of integers
print(f'new {[2*item for item in myList]}') #everything betwen [] is a list comprehension that multiplies each item in the list by 2
#note, this a copy of the list your iterating over, so modifying the new list does not affect the original list
myList
```

### List comprehensions with filters

```python
myList = list(range(100))#creates a list of integers from 0 to 99
filteredList = [item for item in myList if item % 10 == 0]#returns a **NEW** list of items that are multiples of 10
filteredList
```

```python
filteredList = [item for item in myList if item % 10 < 3]
print(filteredList)
```

### List comprehensions with functions

```python
myString = 'My name is Ryan Mitchell. I live in Boston.'
myString.split('.') #splits the string into a list of substrings at each period
#this case you get 2 sentences
#if nothing is possed to split, it splits on spaces
```

```python
myString.split()
```

```python
def cleanWord(word):
    return word.replace('.', '').lower()#removes periods and makes everything lowercase

[cleanWord(word) for word in myString.split()] #same as above but using the list comprehension to call cleanWord on each word
```

```python
[cleanWord(word) for word in myString.split() if len(cleanWord(word)) < 3]
```

### Nested list comprehensions

```python
[[cleanWord(word) for word in sentence.split()] for sentence in myString.split('.')]
#this is a nested list comprehension. iterates over sentences and then words within each sentence
```
